```
import yfinance as yf
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from datetime import datetime, timedelta

# ============================================================
# ส่วนที่ 1: โหลดข้อมูลหุ้น ย้อนหลัง 5 ปี
# ------------------------------------------------------------
# เลือกหุ้น 1 ตัว แล้วดึงข้อมูลราคาย้อนหลัง 5 ปี
# yfinance คืนค่า DataFrame ที่มี column: Open, High, Low,
# Close, Volume — เราจะใช้ Close เป็นหลัก
# ============================================================
ticker     = "AAPL"
end_date   = datetime.now()
start_date = end_date - timedelta(days=5*365)

data = yf.download(ticker, start=start_date, end=end_date)

# บางเวอร์ชัน yfinance ให้ MultiIndex columns → แปลงให้แบน
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

# ============================================================
# ส่วนที่ 2: คำนวณ Daily Return → Annual μ และ Annual σ
# ------------------------------------------------------------
# สูตร Log Return:
#   r_t = ln(Close_t / Close_{t-1})
#   → ใช้ log return เพราะ GBM สมมติ ln(S) เป็น Normal
#   → รับประกันราคาไม่ติดลบ (exp() > 0 เสมอ)
#
# สูตร Annualize (สมมติ 252 วันทำการ/ปี):
#   μ (annual drift)      = mean(r_t) × 252
#   σ (annual volatility) = std(r_t)  × √252
# ============================================================
data['LogReturns'] = np.log(data['Close'] / data['Close'].shift(1))   # r_t = ln(P_t / P_{t-1})
data = data.dropna()

mu_daily    = data['LogReturns'].mean()        # mean รายวัน (log return)
sigma_daily = data['LogReturns'].std()         # std รายวัน  (log return)

mu    = mu_daily    * 252                      # μ annual = μ_daily × 252
sigma = sigma_daily * np.sqrt(252)             # σ annual = σ_daily × √252

S0 = float(data['Close'].iloc[-1])             # ราคาล่าสุด = จุดเริ่มต้น simulate

print("=" * 60)
print(f"  {ticker} — GBM Parameter Summary")
print("=" * 60)
print(f"  Daily  σ = {sigma_daily:.4f}  ← std ของ log return รายวัน")
print(f"  Annual σ = {sigma:.4f}  ← Daily σ × √252  (ความผันผวนต่อปี)")
print(f"  Annual μ = {mu:.4f}  ← mean log return รายวัน × 252 (drift ต่อปี)")
print(f"  S0       = {S0:.2f} USD  ← ราคาปิดล่าสุดของ {ticker}")

# ============================================================
# ส่วนที่ 3: แสดงสมการ GBM, VaR และ CVaR
# ------------------------------------------------------------
# GBM (Geometric Brownian Motion) อธิบายพฤติกรรมราคาหุ้น
#
# รูปแบบ Continuous (SDE):
#   dS/S = μ·dt + σ·dW
#
# รูปแบบ Discrete (Euler-Maruyama — แสดงเพื่อเปรียบเทียบ):
#   ΔS = μ·S·Δt + σ·S·ε·√Δt    โดย ε ~ N(0,1)
#   → เราไม่ได้ใช้วิธีนี้ simulate จริง เพราะมี error สะสมทุก step
#
# VaR 5%  = percentile ที่ 5 ของ price distribution
#   VaR%  = (S₀ - VaR_price) / S₀
#
# CVaR 5% = E[ Price | Price ≤ VaR ]  (Expected Shortfall)
#         = mean ของ prices ที่แย่กว่า VaR ทั้งหมด
#   CVaR% = (S₀ - CVaR_price) / S₀
# ============================================================
print("=" * 60)
print("  สมการ GBM, VaR และ CVaR")
print("=" * 60)
print("  [Continuous SDE]  dS/S = μ·dt + σ·dW")
print(f"    แทนค่า: dS/S = {mu:.4f}·dt + {sigma:.4f}·dW")
print()
print("  [Discrete Step]   ΔS = μ·S·Δt + σ·S·ε·√Δt   (ε ~ N(0,1))")
print(f"    แทนค่า: ΔS = {mu:.4f}·S·{1/252:.5f} + {sigma:.4f}·S·ε·{np.sqrt(1/252):.5f}")
print()
print("  [Exact Solution]  S(t) = S₀·exp( (μ - σ²/2)·t + σ·W(t) )")
print(f"    Ito correction: -σ²/2 = -{0.5*sigma**2:.4f}  (ลดออกจาก drift)")
print("    → exp() รับประกันราคาไม่ติดลบเสมอ")
print()
print("  [VaR 5%]          VaR_price = percentile( prices, 5 )")
print("                    VaR%      = (S₀ - VaR_price) / S₀")
print("    → ขีดล่าง: มีโอกาส 5% ที่ราคาจะต่ำกว่านี้")
print()
print("  [CVaR 5%]         CVaR_price = E[ Price | Price ≤ VaR_price ]")
print("                              = mean( prices ที่ต่ำกว่า VaR_price )")
print("                    CVaR%      = (S₀ - CVaR_price) / S₀")
print("    → ถ้าเกิด worst case จริง: ขาดทุนเฉลี่ยเท่านี้")
print("    → CVaR ≥ VaR เสมอ (เพราะ CVaR คือค่าเฉลี่ยใน tail)")
print("=" * 60)

# ============================================================
# ส่วนที่ 4: ตั้งค่าพารามิเตอร์ Simulation
# ------------------------------------------------------------
# T            = ระยะเวลา simulate (1 ปี)
# dt           = ขนาด 1 step = 1 วันทำการ = 1/252 ปี
# N            = จำนวน steps = 252 (1 ปีมี 252 วันทำการ)
# n_paths      = จำนวน scenarios ที่ simulate (5,000 paths)
# RENDER_PATHS = จำนวน paths ที่วาดบนกราฟ — ใช้แค่ 30 เส้น
#                เพราะวาด 5,000 เส้นพร้อมกัน browser จะช้ามาก
#                30 เส้นก็เพียงพอให้เห็น pattern การกระจายตัว
# STEP         = สร้าง animation frame ทุก 3 วัน (ไม่ใช่ทุกวัน)
#                ลด frame count จาก 252 → ~83 frame
#                animation ยังลื่นอยู่ แต่โหลดเร็วกว่า
# ============================================================
T            = 1.0
dt           = 1/252               # 1 วันทำการ = 1/252 ปี
N            = int(T / dt)         # N = 252 steps
n_paths      = 5000                # จำนวน scenarios ทั้งหมด
RENDER_PATHS = 30                  # paths ที่วาดบนกราฟ (ลดเพื่อ performance)
STEP         = 3                   # สร้าง frame ทุก 3 วัน เพื่อลด frame count

t = np.linspace(0, T, N).reshape(1, -1)   # เวลา shape (1, 252) สำหรับ broadcast

# ============================================================
# ส่วนที่ 5: Simulate GBM ด้วย Exact Solution
# ------------------------------------------------------------
# เราใช้ Exact Solution แทน Discrete Euler-Maruyama เพราะ:
#   - Exact Solution = คำตอบที่แม่นตรงของ Continuous SDE (ได้จาก Ito's Lemma)
#   - Discrete Euler = approximation ที่มี error สะสมทุก step
#
# Exact Solution:
#   S(t) = S₀ · exp( (μ - σ²/2)·t  +  σ·W(t) )
#
#   (μ - σ²/2) คือ Ito correction — ต้องลบ σ²/2 ออกจาก drift
#   เพราะ ln(S) ไม่ใช่ linear ใน S ทำให้ variance กิน drift ไปบางส่วน
#
# Wiener Process W(t):
#   dW ~ N(0, √dt)      ← random increment แต่ละ step (1 วันทำการ)
#   W(0) = 0            ← Brownian motion ต้องเริ่มที่ 0 เสมอ
#   W(t) = Σ dW_i       ← cumulative sum = Brownian motion สะสม
#
#   สำคัญ: ต้อง prepend 0 ก่อน cumsum เพื่อให้ W(0) = 0
#   ถ้าไม่ทำ → W[0] = dW[0] (random) → S[0] ≠ S₀
# ============================================================
np.random.seed(42)   # ตั้ง seed เพื่อให้ผลเหมือนเดิมทุกครั้งที่รัน (reproducibility)

dW = np.random.normal(0, np.sqrt(dt), (n_paths, N-1))  # dW ~ N(0, √dt)  shape: (5000, 251)
W  = np.hstack([np.zeros((n_paths, 1)),                 # W(0) = 0 เสมอ
                np.cumsum(dW, axis=1)])                  # W(t) = Σ dW     shape: (5000, 252)

# S(t) = S₀ · exp( (μ - σ²/2)·t + σ·W(t) )
S = S0 * np.exp(
    (mu - 0.5 * sigma**2) * t     # drift term:     (μ - σ²/2)·t  → shape (1, 252)
    + sigma * W                    # diffusion term: σ·W(t)        → shape (5000, 252)
)                                  # S shape: (5000, 252) — 5000 paths × 252 วัน

# ============================================================
# ส่วนที่ 6: คำนวณ Final VaR 5% และ CVaR 5%
# ------------------------------------------------------------
# VaR_price  = percentile(sorted_prices, 5)
# CVaR_price = mean( prices ≤ VaR_price )   ← Expected Shortfall
# ============================================================
final_prices        = S[:, -1]                          # ราคาวันที่ 252: shape (5000,)
final_prices_sorted = np.sort(final_prices)             # เรียงน้อย → มาก

final_var_price = np.percentile(final_prices_sorted, 5) # VaR = percentile ที่ 5
final_VaR       = (S0 - final_var_price) / S0           # % loss

tail_prices      = final_prices_sorted[final_prices_sorted <= final_var_price]  # worst 5%
final_cvar_price = float(np.mean(tail_prices))          # CVaR = ค่าเฉลี่ยของ tail
final_CVaR       = (S0 - final_cvar_price) / S0         # % loss เฉลี่ยของ worst case

print("=" * 60)
print("  Final VaR 5%  (Value at Risk)")
print("=" * 60)
print(f"  สูตร: VaR_price = percentile( 5,000 prices, 5 )")
print(f"        VaR%      = (S₀ - VaR_price) / S₀")
print(f"  VaR Price = {final_var_price:.2f} USD")
print(f"  VaR %Loss = {final_VaR*100:.2f}%")
print(f"  → มีโอกาส 5% ที่ราคาจะต่ำกว่า {final_var_price:.2f} USD")
print("=" * 60)
print()
print("=" * 60)
print("  Final CVaR 5%  (Expected Shortfall)")
print("=" * 60)
print(f"  สูตร: CVaR_price = mean( prices ที่ ≤ VaR_price )")
print(f"        CVaR%      = (S₀ - CVaR_price) / S₀")
print(f"  CVaR Price = {final_cvar_price:.2f} USD  (จาก {len(tail_prices)} paths)")
print(f"  CVaR %Loss = {final_CVaR*100:.2f}%")
print(f"  → ถ้าเกิด worst case จริง ขาดทุนเฉลี่ย {final_CVaR*100:.2f}%")
print()
print(f"  เปรียบเทียบ:")
print(f"  VaR  = {final_var_price:.2f} USD ({final_VaR*100:.1f}%)  ← เส้นแบ่ง worst 5%")
print(f"  CVaR = {final_cvar_price:.2f} USD ({final_CVaR*100:.1f}%)  ← เฉลี่ยของ worst 5%")
print("=" * 60)

# ============================================================
# ส่วนที่ 6b: lock x-axis ของ histogram ให้คงที่ตลอด animation
# ------------------------------------------------------------
# ใช้ percentile 1–99 แทน min/max เพื่อตัด outlier สุดขั้ว
# ถ้าใช้ min/max → x-axis จะกว้างมากเพราะ outlier ดึงออกไป
# ============================================================
x_min, x_max = np.percentile(S.flatten(), [1, 99])

# ------------------------------------------------------------
# คำนวณสีของแต่ละ path ตามราคาสุดท้าย (Day 252)
# input : final_price — ราคาวันที่ 252 ของ path นั้น
# output: rgba string — สีของเส้น
#   ราคาต่ำสุด (norm=0.0) → แดง   rgba(220, 0, 50)
#   ราคากลาง  (norm=0.5) → เหลือง rgba(220,180, 50)
#   ราคาสูงสุด (norm=1.0) → เขียว rgba(  0,190, 50)
# ------------------------------------------------------------
_p_min = S[:RENDER_PATHS, -1].min()
_p_max = S[:RENDER_PATHS, -1].max()

def _path_color(final_price, alpha=0.6):
    """คืนค่าสี rgba ตาม final_price เทียบกับ min/max ของ 30 paths"""
    norm = (final_price - _p_min) / (_p_max - _p_min + 1e-9)  # normalize 0–1
    if norm < 0.5:                                              # ต่ำกว่ากลาง → แดง→เหลือง
        r, g, b = 220, int(180 * norm * 2), 50
    else:                                                       # สูงกว่ากลาง → เหลือง→เขียว
        r, g, b = int(220 * (1 - (norm - 0.5) * 2)), 190, 50
    return f'rgba({r},{g},{b},{alpha})'

path_colors = [_path_color(S[i, -1]) for i in range(RENDER_PATHS)]

# คำนวณ Max / Min / Avg ของทุก path รายวัน (จาก 5,000 paths ทั้งหมด)
S_max = S.max(axis=0)    # ราคาสูงสุดแต่ละวัน  shape (252,)
S_min = S.min(axis=0)    # ราคาต่ำสุดแต่ละวัน  shape (252,)
S_avg = S.mean(axis=0)   # ราคาเฉลี่ยแต่ละวัน  shape (252,)

print("=" * 60)
print("  Price Range after Simulation (Day 252, 5,000 paths)")
print("=" * 60)
print(f"  Max Price = {S_max[-1]:.2f} USD  ← ราคาสูงสุดจากทุก path")
print(f"  Avg Price = {S_avg[-1]:.2f} USD  ← ราคาเฉลี่ยจากทุก path")
print(f"  Min Price = {S_min[-1]:.2f} USD  ← ราคาต่ำสุดจากทุก path")
print("=" * 60)

# ============================================================
# ส่วนที่ 7: สร้าง Animation Frames ทีละวัน
# ------------------------------------------------------------
# แต่ละ frame k แสดงสถานะของ simulation ณ วันที่ k
#
# สูตร VaR 5% (Direct Price Percentile):
#   prices_sorted = sort( S[:, k] )          ← เรียง 5000 ราคา ณ วันที่ k
#   VaR_price     = percentile(sorted, 5)    ← ราคาต่ำสุด 5%
#   VaR_pct       = (S₀ - VaR_price) / S₀   ← % loss
#
# สูตร CVaR 5% (Expected Shortfall):
#   tail          = prices ที่ต่ำกว่า VaR_price
#   CVaR_price    = mean(tail)               ← ราคาเฉลี่ยของ worst 5%
#   CVaR_pct      = (S₀ - CVaR_price) / S₀  ← % loss เฉลี่ยของ worst case
#
# แต่ละ frame ประกอบด้วย:
#   - 30 price paths (ฝั่งซ้าย)
#   - เส้นแนวนอน VaR (เหลือง) + CVaR (แดง) ฝั่งซ้าย
#   - histogram ของ price distribution (ฝั่งขวา)
#   - เส้นแนวตั้ง VaR + CVaR (ฝั่งขวา)
# ============================================================
frames = []

for k in range(5, N, STEP):
    prices        = S[:, k]             # ราคาทุก path ณ วันที่ k: shape (5000,)
    prices_sorted = np.sort(prices)     # เรียงน้อย → มาก

    var_price = np.percentile(prices_sorted, 5)
    VaR_pct   = (S0 - var_price) / S0

    tail_k     = prices_sorted[prices_sorted <= var_price]
    cvar_price = float(np.mean(tail_k)) if len(tail_k) > 0 else var_price
    CVaR_pct   = (S0 - cvar_price) / S0

    hist_y, hist_x = np.histogram(prices, bins=40, density=True)
    hist_x_mid     = (hist_x[:-1] + hist_x[1:]) / 2
    max_density    = hist_y.max()

    # แดงเข้ม = CVaR zone (worst of worst) | ส้ม = VaR zone | น้ำเงิน = safe
    bar_colors = []
    for x in hist_x_mid:
        if x <= cvar_price:
            bar_colors.append('rgba(200,50,50,0.9)')
        elif x <= var_price:
            bar_colors.append('rgba(230,120,30,0.85)')
        else:
            bar_colors.append('rgba(65,125,195,0.65)')

    hist = go.Bar(
        x=hist_x_mid, y=hist_y,
        xaxis="x2", yaxis="y2",
        marker_color=bar_colors,
        marker_line=dict(width=0.3, color='rgba(255,255,255,0.4)'),
        showlegend=False
    )

    var_line = go.Scatter(
        x=[var_price, var_price], y=[0, max_density],
        mode="lines+text",
        text=["", f"VaR ${var_price:.1f}<br>({VaR_pct*100:.1f}%)"],
        textposition="top center",
        textfont=dict(color="orange", size=10, family="monospace"),
        line=dict(color="orange", width=2.5),
        name="VaR 5%", xaxis="x2", yaxis="y2",
        showlegend=(k == 5), legendgroup="var"
    )

    cvar_line = go.Scatter(
        x=[cvar_price, cvar_price], y=[0, max_density * 0.6],
        mode="lines+text",
        text=["", f"CVaR ${cvar_price:.1f}<br>({CVaR_pct*100:.1f}%)"],
        textposition="top center",
        textfont=dict(color="rgba(255,100,100,1)", size=10, family="monospace"),
        line=dict(color="rgba(255,100,100,1)", width=2, dash="dash"),
        name="CVaR 5%", xaxis="x2", yaxis="y2",
        showlegend=(k == 5), legendgroup="cvar"
    )

    var_hline = go.Scatter(
        x=[0, t.flatten()[k - 1]], y=[var_price, var_price],
        mode="lines+text",
        text=["", f"  VaR ${var_price:.1f} ({VaR_pct*100:.1f}%)"],
        textposition="middle right",
        textfont=dict(color="orange", size=10),
        line=dict(color="orange", width=2),
        showlegend=False
    )

    cvar_hline = go.Scatter(
        x=[0, t.flatten()[k - 1]], y=[cvar_price, cvar_price],
        mode="lines+text",
        text=["", f"  CVaR ${cvar_price:.1f} ({CVaR_pct*100:.1f}%)"],
        textposition="middle right",
        textfont=dict(color="rgba(255,100,100,1)", size=10),
        line=dict(color="rgba(255,100,100,1)", width=2, dash="dash"),
        showlegend=False
    )

    path_traces = [
        go.Scatter(
            x=t.flatten()[:k], y=S[i][:k],
            mode="lines",
            line=dict(width=0.9, color=path_colors[i]),
            showlegend=False
        )
        for i in range(RENDER_PATHS)
    ]

    tx = t.flatten()[:k]

    max_trace = go.Scatter(x=tx, y=S_max[:k], mode="lines",
        line=dict(color="rgba(60,200,100,0.85)", width=1.5, dash="dot"),
        name="Max", legendgroup="stat", showlegend=(k == 5))
    avg_trace = go.Scatter(x=tx, y=S_avg[:k], mode="lines",
        line=dict(color="rgba(255,255,255,0.95)", width=2),
        name="Avg", legendgroup="stat", showlegend=(k == 5))
    min_trace = go.Scatter(x=tx, y=S_min[:k], mode="lines",
        line=dict(color="rgba(200,80,80,0.85)", width=1.5, dash="dot"),
        name="Min", legendgroup="stat", showlegend=(k == 5))

    frames.append(go.Frame(
        data=path_traces + [var_hline, cvar_hline, max_trace, avg_trace, min_trace, hist, var_line, cvar_line],
        name=str(k)
    ))

# ============================================================
# ส่วนที่ 8: สร้าง Figure, Annotation และ Layout
# ============================================================
fig = go.Figure(data=frames[0].data, frames=frames)

fig.update_layout(
    title=dict(
        text=(
            f"{ticker} — GBM Monte Carlo Simulation  |  "
            f"VaR 5% = {final_var_price:.2f} USD ({final_VaR*100:.1f}%)  |  "
            f"CVaR 5% = {final_cvar_price:.2f} USD ({final_CVaR*100:.1f}%)"
        ),
        font=dict(size=14), x=0.5, xanchor="center"
    ),

    annotations=[
        dict(
            text=(
                "── Parameters ──<br>"
                f"μ  = {mu:.4f} /yr<br>"
                f"σ  = {sigma:.4f}<br>"
                f"S0 = {S0:.2f} USD<br>"
                "Paths=5,000 | Steps=252<br>"
                "<br>"
                "── Price Range Day252 ──<br>"
                f"Max = {S_max[-1]:.2f} USD<br>"
                f"Avg = {S_avg[-1]:.2f} USD<br>"
                f"Min = {S_min[-1]:.2f} USD<br>"
                "<br>"
                "── VaR 5% ──<br>"
                "percentile(prices, 5)<br>"
                f"Price = {final_var_price:.2f} USD<br>"
                f"Loss  = {final_VaR*100:.2f}%<br>"
                "<br>"
                "── CVaR 5% ──<br>"
                "mean(prices ≤ VaR)<br>"
                f"Price = {final_cvar_price:.2f} USD<br>"
                f"Loss  = {final_CVaR*100:.2f}%<br>"
                "<br>"
                "── GBM Equations ──<br>"
                "dS/S = μdt + σdW<br>"
                "S(t) = S₀·exp((μ-σ²/2)t+σW)"
            ),
            xref="paper", yref="paper",
            x=0.91, y=1.0,
            xanchor="left", yanchor="top",
            showarrow=False, align="left",
            bgcolor="rgba(255,255,255,0.88)",
            bordercolor="rgba(150,160,180,0.8)",
            borderwidth=1,
            font=dict(size=10, family="monospace")
        ),
        dict(
            text="🟢 high return  🟡 mid  🔴 low return  |  🟠 VaR 5%  🔴 CVaR 5% (Expected Shortfall)",
            xref="paper", yref="paper",
            x=0.5, y=-0.08,
            xanchor="center", yanchor="top",
            showarrow=False,
            font=dict(size=10, color="rgba(80,80,100,1)")
        ),
    ],

    paper_bgcolor='rgba(245,248,255,1)',
    plot_bgcolor ='rgba(235,241,252,1)',

    xaxis =dict(domain=[0, 0.60], title="Time (years)", gridcolor='rgba(200,210,230,0.6)'),
    yaxis =dict(title="Price (USD)",                    gridcolor='rgba(200,210,230,0.6)'),
    xaxis2=dict(domain=[0.65, 0.89], range=[x_min, x_max], title="Price", gridcolor='rgba(200,210,230,0.6)'),
    yaxis2=dict(anchor="x2", title="Density",           gridcolor='rgba(200,210,230,0.6)'),

    updatemenus=[{
        "type": "buttons",
        "buttons": [
            {"label": "▶ Play",  "method": "animate",
             "args": [None, {"frame": {"duration": 40, "redraw": True}, "fromcurrent": True}]},
            {"label": "⏸ Pause", "method": "animate",
             "args": [[None], {"frame": {"duration": 0}, "mode": "immediate"}]}
        ]
    }],

    sliders=[{
        "steps": [
            {"method": "animate", "label": str(k),
             "args": [[str(k)], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}]}
            for k in range(5, N, STEP)
        ],
        "currentvalue": {"prefix": "Day: "}
    }]
)

fig.show()
```
